```python
import stable_retro as retro
import gymnasium as gym
from stable_baselines3 import PPO
from stable_baselines3.common.vec_env import DummyVecEnv
from stable_baselines3.common.evaluation import evaluate_policy
from stable_baselines3.common.monitor import Monitor
from stable_baselines3.common.vec_env import VecFrameStack, VecVideoRecorder
import os
from stable_baselines3.common.callbacks import CheckpointCallback
```

```python
game_name = "StreetFighterIISpecialChampionEdition-Genesis-v0"
log_path = os.path.join("Training", "Logs")
```

```python
del env
```

```python
env = DummyVecEnv([lambda: Monitor(retro.make(game=game_name, render_mode="None"))])
env = VecFrameStack(env, n_stack=4)
```

```python
model = PPO("CnnPolicy", env, verbose=1, device="mps", tensorboard_log=log_path)
checkpoint = CheckpointCallback(
    save_freq=100_000,
    save_path="./Training/Saved Models/Checkpoints",
    name_prefix="PPO_sf",
)
```

```python
model.learn(total_timesteps=1000000, callback=checkpoint, progress_bar=True)
```

```python
PPO_path = os.path.join("Training", "Saved Models", "PPO_Model_SF")
```

```python
model.save(PPO_path)
```

```python
del model
```

```python
model = PPO.load(PPO_path, env)
```

```python
env.close()
```

```python
env_eval = DummyVecEnv([lambda: Monitor(retro.make(game=game_name, render_mode="human"))])
env_eval = VecFrameStack(env_eval, n_stack=4)
```

```python
evaluate_policy(model, env_eval, n_eval_episodes=10)
```

```python
env_record = DummyVecEnv([lambda: Monitor(retro.make(game=game_name, render_mode="rgb_array"))])
env_record = VecFrameStack(env_record, n_stack=4)

env_record = VecVideoRecorder(
    env_record,
    video_folder="./Training/Recordings",
    record_video_trigger=lambda step: step == 0,
    name_prefix="SF_Test",
    video_length=5000
)

evaluate_policy(model, env_record, n_eval_episodes=1)
env_record.close()
```
